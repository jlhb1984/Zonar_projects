Conversor de temperatura de Farenheit a Celsius

```python
# Entrada de datos
fahrenheit = float(input("Ingrese la temperatura en Fahrenheit (°F): "))

# Proceso de conversión
celsius = (fahrenheit - 32) * (5 / 9)

# Salida de resultados
print("La temperatura equivalente en Celsius es:", celsius, "°C")
```

Menú variado con 4 ingredientes

```python
ingredientes = ["Tomate", "Lechuga", "Sardinas", "Atún"]
total = len(ingredientes)
numero_menu = 0

print("=== LISTA DE MENÚS VARIADOS (Combinaciones de 2 ingredientes) ===")

# Bucle externo: recorre desde el primer ingrediente hasta el penúltimo
for i in range(total - 1):
    
    # Bucle interno: recorre los ingredientes siguientes al índice 'i'
    for j in range(i + 1, total):
        
        numero_menu += 1
        
        # Imprimir el menú resultante de la combinación actual
        print(f"Menú #{numero_menu}: {ingredientes[i]} y {ingredientes[j]}")

print(f"Total de menús únicos generados: {numero_menu}")
```

Conversor de divisas

```python
print("=========================================")
print("   CONVERSOR DE DIVISAS: USD <-> COP     ")
print("=========================================")
print("Seleccione la dirección de la conversión:")
print("1. Dólares (USD) a Pesos Colombianos (COP)")
print("2. Pesos Colombianos (COP) a Dólares (USD)")

# Entrada de la opción seleccionada
opcion = int(input("Ingrese una opción (1 o 2): "))

# Validar y ejecutar según la opción seleccionada
if opcion == 1:
    print("--- Conversión de USD a COP ---")
    cantidad = float(input("Ingrese el monto en Dólares (USD): "))
    tasaCambio = float(input("Ingrese la tasa de cambio actual (valor de 1 USD en COP): "))

    if cantidad >= 0 and tasaCambio > 0:
        resultado = cantidad * tasaCambio
        print("-----------------------------------------")
        print(f"Resultado: ${cantidad} USD equivalen a ${resultado} COP")
    else:
        print("Error: Los valores ingresados deben ser mayores a cero.")

elif opcion == 2:
    print("--- Conversión de COP a USD ---")
    cantidad = float(input("Ingrese el monto en Pesos Colombianos (COP): "))
    print("Ingrese la tasa de cambio actual (valor de 1 USD en COP):")
    tasaCambio = float(input())

    if cantidad >= 0 and tasaCambio > 0:
        resultado = cantidad / tasaCambio
        print("-----------------------------------------")
        print(f"Resultado: ${cantidad} COP equivalen a ${resultado} USD")
    else:
        print("Error: Los valores ingresados deben ser mayores a cero.")

else:
    print("Opción inválida. Por favor, ejecute el programa nuevamente.")

print("=========================================")
print("Proceso finalizado con éxito.")
```
