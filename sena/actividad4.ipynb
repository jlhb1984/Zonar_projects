```python
filtro_menos_60=df_completo[df_completo['Edad']<60 ]
filtro_mas_15=filtro_menos_60[filtro_menos_60['Edad']>18]
#filtro_mas_15.info()
filtro_mas_15['Edad'].value_counts(ascending=True)
df=filtro_mas_15

df.info()
```

```python
df['Genero'].value_counts()
```

```python
df['Color'].value_counts()
```

```python
df.Color=df.Color.str.upper()
df['Color'].value_counts()
```

```python
df=df.drop(['Unnamed: 0'],axis=1)
df.info()
```

```python
df['Total']=0.0
for i in range(0,df.shape[0]):
    df.iloc[i,9]=df.iloc[i,3]*df.iloc[i,4]
df.head(3)
```

```python
sns.catplot(x='Edad',data=df,kind='box',aspect = 1.5)
```

```python
df['Rango']='NaN'
for i in range(0,df.shape[0]):
    if df.iloc[i,6]>18 and df.iloc[i,6]<28:
        df.iloc[i,10]='a'
    if df.iloc[i,6]>39 and df.iloc[i,6]<48:
            df.iloc[i,10]='b'
    if df.iloc[i,6]>49 and df.iloc[i,6]<60:
            df.iloc[i,10]='c'
df.head(5)           
```

```python
df.groupby(by='Rango').sum()
```
