Speed

```python
v=float(input("Type the value in km/h: "))
w=(v*100000)/3600
print(v,'km/h',' -> ',w,'cm/s')
```

```python
w=float(input("Type the value in cm/s: "))
v=(w*3600)/100000
x=v*0.62137119
print(w,'cm/s',' -> ',v,'km/h',' -> ',x,'mph')
```

```python
y=float(input("Type the value in mph: "))
z=y*1.609344
print(y,'mph',' -> ',z,'km/h')
```

El sistema reportará alertas a 80Km/h si se establece a 82Km/h porque la variable es entera y aproxima al entero menor.

Volume

```python
t=float(input("Type the value in gallons: "))
u=t*3.785412
print(t,'galons',' -> ',u,'liters')
```

```python
u=float(input("Type the value in liters: "))
t=u/3.785412
print(u,'liters',' -> ',t,'galons')
```

Temperature

```python
c=float(input("Type the Celsius value: "))
f=(9/5)*c+32
print(c,'Celsius',' -> ',f,'Farenheit')
```

```python
f=float(input("Type the Farenheit value: "))
c=(f-32)*(5/9)
print(f,'Farenheit',' -> ',c,'celsius')
```
