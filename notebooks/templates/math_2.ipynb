```python
pi=math.pi
an=[]
bn=[]
fn=[]
T=1

for i in range(0,10):
    fn.append( 2/((1-4*(i**2))*pi) )

t=np.linspace(0,10,1000)

print(fn)

y1=((fn[0]*np.cos(2*pi*t*0)+fn[1]*np.cos(2*pi*t*1)+fn[2]*np.cos(2*pi*t*2)+fn[3]*np.cos(2*pi*t*3)+fn[4]*np.cos(2*pi*t*4)))

plt.plot(t,y1)

plt.show()
```
