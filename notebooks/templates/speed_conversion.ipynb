```python
v=float(input("Type the value in km/h: "))
w=(v*100000)/3600
print(v,'km/h',' -> ',w,'cm/s')
```

```python
w=float(input("Type the value in cm/s: "))
v=(w*3600)/100000
x=v*0.62137119
print(w,'cm/s',' -> ',v,'km/h',' -> ',x,'mph')
```

```python
y=float(input("Type the value in mph: "))
z=y*1.609344
print(y,'mph',' -> ',z,'km/h')
```
