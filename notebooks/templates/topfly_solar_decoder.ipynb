```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
raw_data=pd.read_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\zonar\solar_daniel_g.csv',sep=',')
headers=[]
inner_battery_v=[]
date=[]
raw=[]
unit=pd.DataFrame()

for i in range(0,raw_data.shape[0]):
    headers.append(raw_data.iloc[i,28][0:4])
    inner_battery_v.append(raw_data.iloc[i,28][88:90])
    date.append(raw_data.iloc[i,28][34:46])
    raw.append(raw_data.iloc[i,28])

unit['Headers']=headers
unit['inner_battery']=inner_battery_v
unit['date']=date
unit['raw']=raw

unit_sorted=unit.sort_values('date',ascending=True)

unit_sorted.info()
```

```python
x=[]
batt=[]
for i in range(0,unit_sorted.shape[0]):
    aux_batt=int(unit_sorted.iloc[i,1])
    batt.append(aux_batt)
    if aux_batt<20:
        print(unit_sorted.iloc[i,1],unit_sorted.iloc[i,2],unit_sorted.iloc[i,3])
    x.append(i)

plt.plot(x,batt)
plt.ylabel('Battery %')
plt.xlabel("Sample number")
```
