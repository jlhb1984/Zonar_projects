```python
import pandas as pd
import requests
```

MRR for customers from Colombia.

```python
customer_col=pd.read_excel(r'C:\Users\Jose.Hurtado\Downloads\CustomerUnits.xlsx')

url = "https://www.datos.gov.co/resource/ceyp-9c7c.json?$limit=1&$order=vigenciadesde DESC"
response = requests.get(url)

data = response.json()
trm = float(data[0]["valor"])

aux=0

for i in range(0,customer_col.shape[0]):
    aux=aux+customer_col.iloc[i,10]
aux=aux/trm

print("Total MRR in USD: ", aux)
```
