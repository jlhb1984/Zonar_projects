```python
# Width of 1s for both pulses.

t=np.linspace(0,10000,100000)

an1=[]
bn1=[]
an2=[]
bn2=[]
T1=105
T2=255
a01=1/T1
a02=1/T2
w1=(2*pi)/T1
w2=(2*pi)/T2

for n in range(1,1000):
    an1.append((1/(n*pi))*np.sin((2*pi*n)/T1))
    bn1.append((1/(n*pi))*(1-np.cos((2*pi*n)/T1)))
    #print(an1,bn1)

for n in range(1,1000):
    an2.append((1/(n*pi))*np.sin((2*pi*n)/T2))
    bn2.append((1/(n*pi))*(1-np.cos((2*pi*n)/T2)))
    #print(an2,bn2)

aux=0

for i in range(1,1000):
    y1=(an1[i-1]*np.cos(w1*t*i)*1+bn1[i-1]*np.sin(w1*t*i)) +aux
    aux=y1

aux=0
for i in range(1,1000):
    y2=(an2[i-1]*np.cos(w2*t*i)*1+bn2[i-1]*np.sin(w2*t*i)) +aux
    aux=y2

y1=a01+y1
y2=a02+y2

plt.plot(t,y1)
plt.plot(t,y2+1.5)
plt.xlim(0,2000)
```
