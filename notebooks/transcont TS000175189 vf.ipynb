```python
vf_date_time=vf[vf['Message'].str.contains('AutoSnapshot')]
aux_date_time=[]
epoch_date=[]
speed=[]
auto_snapshot=[]
vf_total=pd.DataFrame()

for i in range(0,vf_date_time.shape[0]):
    aux_date_time.append(float(vf_date_time.iloc[i,3][5:18])/1000)
    epoch_date.append(datetime.datetime.fromtimestamp(aux_date_time[i]))
    speed.append(vf_date_time.iloc[i,12][6:18])
    auto_snapshot.append(vf_date_time.iloc[i,5])

vf_total['epoch']=aux_date_time
vf_total['datetime']=epoch_date
vf_total['speed']=speed
vf_total['autosnapshot']=auto_snapshot

vf_total_filtered=vf_total[vf_total['epoch']>=1.784857e+09]
vf_total_filtered.to_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\transcont\auto_snapshot.csv', index=False)
```
