```python
aux=[]
aux_miles=[]
x_axis=[]
date=[]
ett651_odometer=pd.DataFrame()
cont=0

for i in range (0, ett651.shape[0]):
    l=len(ett651.iloc[i,2])
    #print('Len=',l)
    if l<180:
        aux.append(int(ett651.iloc[i,2][68:76],16))
        aux_miles.append((int(ett651.iloc[i,2][68:76],16))*0.621371)
        date.append(ett651.iloc[i,0])
        cont=cont+1 

ett651_odometer['Odometer in meters']=aux
ett651_odometer['Odometer en miles']=aux_miles
ett651_odometer['Date']=date

for i in range(0,cont):
    x_axis.append(i)

ett651_odometer
plt.plot(x_axis,aux)
```

```python
print('Min: ',ett651_odometer['Odometer in meters'].min(),'Max: ',ett651_odometer['Odometer in meters'].max(),'Diff= ',ett651_odometer['Odometer in meters'].max()-ett651_odometer['Odometer in meters'].min())
```
