```python
vf_autosnapshot=vf[vf['Unnamed: 3'].str.contains('AutoSnapshot')]
time_ign_on=[]
time_ign_off=[]
autosnapshot=[]
vf_autosnapshot_time=pd.DataFrame()

for i in range(0,vf_autosnapshot.shape[0]):
    aux=float(vf_autosnapshot.iloc[i,7][19:33])/1000
    time_ign_on.append(datetime.datetime.fromtimestamp(aux))
    aux=float(vf_autosnapshot.iloc[i,8][20:34])/1000
    time_ign_off.append(datetime.datetime.fromtimestamp(aux))
    autosnapshot.append(vf_autosnapshot.iloc[i,11][10:])

vf_autosnapshot_time['last_time_ign_on']=time_ign_on
vf_autosnapshot_time['last_time_ign_off']=time_ign_off
vf_autosnapshot_time['URL']=autosnapshot

vf_autosnapshot_time.to_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\cyc\bvk799_autos.csv')

#vf_autosnapshot[['Unnamed: 11','Unnamed: 7','Unnamed: 8']].to_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\chilco\qor080 vf 20-22 autoSnapShot.csv')
```

```python
ign_on=[]
ign_off=[]
ign_on_col=[]
ign_off_col=[]
ign_time=pd.DataFrame()

vf_camera_events_ignition=vf[vf['Unnamed: 3'].str.contains('Ignition')]
for i in range(0,vf_camera_events_ignition.shape[0]):

    ign_on.append(vf_camera_events_ignition.iloc[i,8][19:33])
    aux=float(vf_camera_events_ignition.iloc[i,8][19:33])/1000
    ign_on_col.append(datetime.datetime.fromtimestamp(aux))

    ign_off.append(vf_camera_events_ignition.iloc[i,9][20:33])
    aux=float(vf_camera_events_ignition.iloc[i,9][20:34])/1000
    ign_off_col.append(datetime.datetime.fromtimestamp(aux))    

ign_time['epoch_ign_on']=ign_on
ign_time['epoch_ign_on_col']=ign_on_col
ign_time['epoch_ign_off']=ign_off
ign_time['epoch_ign_off_col']=ign_off_col

ign_time.to_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\cyc\bvk799_ig.csv')
```
