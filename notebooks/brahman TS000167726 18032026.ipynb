```python
import pandas as pd
import numpy as np
```

```python
#ars in Kilometers
ars_12022026=pd.read_csv(r'C:\Users\Jose.Hurtado\Documents\projects\data\brahman\Activity Report Summarized_es_2026-03-18.csv',sep=',')
ars_12022026['MilesDriven_in_Miles']=0.0
cad_str_to_float=[]

for i in range(0,ars_12022026.shape[0]):
    aux=ars_12022026.iloc[i,4].replace(',','.')
    cad_str_to_float.append(aux)

ars_12022026['MilesDriven_dec']=cad_str_to_float
ars_12022026['MilesDriven_dec']

ars_12022026.info()

for i in range(0,ars_12022026.shape[0]):
    ars_12022026.iloc[i,18]=float(ars_12022026.iloc[i,19])*0.621371
    #print(ars_12022026.iloc[i,19],ars_12022026.iloc[i,18])
```

```python
#ur in miles
ur_12022026=pd.read_csv(r'C:\Users\Jose.Hurtado\Documents\projects\data\brahman\utilizationReport_2026-03-18.csv',sep=',')
ur_12022026.info()
```

```python
value_to_look=input('Type the short name to look for: ')
ars_12022026_sn=ars_12022026[ars_12022026['UnitName1'].str.contains(value_to_look)]
ars_12022026_sn[['UnitName1','MilesDriven','MilesDriven_in_Miles']]
```

```python
value_to_look=input('Type the short name to look for: ')
ur_12022026_sn=ur_12022026[ur_12022026['Unit'].str.contains(value_to_look)]
ur_12022026_sn[['Unit','Distance (mi)',]]
```

```python
value_to_look=input('Type the short name to look for: ')
ur_12022026_sn=ur_12022026[ur_12022026['Unit'].str.contains(value_to_look)]
ars_12022026_sn=ars_12022026[ars_12022026['UnitName1'].str.contains(value_to_look)]
print("Short name typed: ", value_to_look)
print("\n")
print('Actividades (Sumarizado) from TSO:')
print(ars_12022026_sn[['UnitName1','MilesDriven','MilesDriven_in_Miles']])
print("\n")
print('Utilization Report from Ignition:')
print(ur_12022026_sn[['Unit','Distance (mi)',]])
```
