```python
import pandas as pd
import numpy as np
```

```python
df_book_table01_aux=pd.read_excel(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\transcarga\CustomerUnits.xlsx')
df_book_table02_aux=pd.read_excel(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\transcarga\message logs 13_6_2026, 9_38_00 a. m. - 16_6_2026, 12_00_00 p. m..xlsx')
df_book_table02_aux['phone']='Nan'

for i in range(0, df_book_table02_aux.shape[0]):
    df_book_table02_aux.iloc[i,17]=str(df_book_table02_aux.iloc[i,6])[2:12]

df_book_table01_aux.info()
df_book_table02_aux.info()

#df_book_table02_aux['Phone']
df_book_table01=df_book_table01_aux[['Unit','Serial Number','Phone Number','Last Contact On','Unit Type','Carrier','SIM #']]
df_book_table02=df_book_table02_aux[['text','status','phone','doneAt']]
```

```python
df_book_table01.info()
df_book_table02.info()
```

```python
num_filas_table01=int(df_book_table01.shape[0])
num_filas_table02=int(df_book_table02.shape[0])       
cont=0
df_aux=pd.DataFrame()        
list_field01=[]
list_field02=[]
list_field03=[]
list_field04=[]
list_field05=[]   
list_field06=[]
list_field07=[]
list_field08=[]    
list_field09=[]   
list_field10=[]
#print("df_book_table01[8,0]: ",df_book_table01.iloc[8,0]) #UnitsByDealer.csv_r.csv
#print("df_book_table02[8,5]: ",df_book_table02.iloc[8,5]) #6MB.        
print("\nInformación de las tablas: ")
print("\nTabla 1:")
print(df_book_table01.info())
print("\nTabla 2:")
print(df_book_table02.info())
attribute_name_table01='Phone Number'#Digitar el índice de la tabla 1.#input("\nDigita el atributo a comparar de la tabla 1: ")
attribute_name_table02='phone'#Digitar el índice de la tabla 2.#input("Digita el atribito a comparar de la tabla 2: ")
position_table01=int(df_book_table01.columns.get_loc(attribute_name_table01))
position_table02=int(df_book_table02.columns.get_loc(attribute_name_table02))
print("\nAtributo de la tabla1: ",position_table01)
print("Atributo de la tabla2: ",position_table02,"\n")
print("Trabajando...")                          
for i in range(0,num_filas_table01):
    for j in range(0,num_filas_table02):
        #if str(df_book_table01.iloc[i,5])==str(df_book_table02.iloc[j,3]): 
        if str(df_book_table01.iloc[i,position_table01])==str(df_book_table02.iloc[j,position_table02]):                                                                             
            list_field01.append(df_book_table01.iloc[i,0])
            list_field02.append(df_book_table01.iloc[i,1])
            list_field03.append(df_book_table01.iloc[i,2])
            list_field04.append(df_book_table01.iloc[i,3])
            list_field05.append(df_book_table02.iloc[j,0])
            list_field06.append(df_book_table02.iloc[j,1])
            list_field07.append(df_book_table02.iloc[j,3])
            list_field08.append(df_book_table01.iloc[i,4])
            list_field09.append(df_book_table01.iloc[i,5])
            list_field10.append(df_book_table01.iloc[i,6])
            print("df_book_table01[i,position_table01]: ",df_book_table01.iloc[i,position_table01]," = ",end="")
            print("df_book_table02[j,position_table02]: ",df_book_table02.iloc[j,position_table02])               
            cont=cont+1
df_aux['Unit']=list_field01
df_aux['Serial Number']=list_field02
df_aux['Phone']=list_field03
df_aux['Last contact on']=list_field04
df_aux['text']=list_field05
df_aux['status']=list_field06
df_aux['date']=list_field07
df_aux['Unit Type']=list_field08
df_aux['Carrier']=list_field09
df_aux['Sim #']=list_field10

print("Cont= ",cont)
print(df_aux.info())
option_exp=input("\n¿Desea generar un reporte de coincidencias en formato csv S/N?\n")
if option_exp=="S":
    df_aux.to_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\transcarga\coincidences.csv')                  #Edit
cont=0
option_exp=input("¿Desea generar un reporte de diferencias en formato csv S/N?\n")        
df_book_table02['ACTION']="Keep"
if option_exp=="S":
    print("Trabajando...")
    for i in range(0,num_filas_table01):
        for j in range(0,num_filas_table02):
            if str(df_book_table01.iloc[i,position_table01])==str(df_book_table02.iloc[j,position_table02]):
                df_book_table02.iloc[j,3]=np.nan
                print("df_book_table01[i,position_table01]: ",df_book_table01.iloc[i,position_table01]," = ",end="")
                print("df_book_table02[j,position_table02]: ",df_book_table02.iloc[j,position_table02])               
    #print(df_book_table02)
    df_book_table02.dropna(axis=0,inplace=True)
    print(df_book_table02)
    df_book_table02.to_csv(r'C:\Users\Jose.Hurtado\Documents\Zonar_projects\data\transcarga\differences.csv')          #Edit

#lista_combinada=[lista_simcard,lista_linea]
```
